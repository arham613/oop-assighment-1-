**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


# Usage
s = Student("Ali", 101)
print(s.get_name(), s.get_roll_number())

s.set_name("Ahmed")
s.set_roll_number(102)
print(s.get_name(), s.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance or invalid amount!")

    def get_balance(self):
        return self.__balance


# Usage
account = BankAccount(100)
account.deposit(50)
account.withdraw(200)  # Rejection: balance won't fall below zero
account.withdraw(80)
print("Balance:", account.get_balance())
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
class Employee:
    def __init__(self, salary):
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary


# Usage
emp = Employee(50000)
emp.set_salary(-1000)  # Prints: Invalid Salary[cite: 1]
emp.set_salary(60000)
print("Salary:", emp.get_salary())
```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
# Parent Class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

# Child Class
class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)  # Parent class ka constructor call karta hai
        self.model = model

# Car ka object create karna
my_car = Car("Toyota", "Corolla")

# Both attributes print karna
print("Brand:", my_car.brand)
print("Model:", my_car.model)
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Parent Class 1
class Animal:
    def eat(self):
        print("Animal kha raha hai")

# Parent Class 2 (Inherits Animal)
class Mammal(Animal):
    def walk(self):
        print("Mammal chal sakta hai")

# Child Class (Inherits Mammal)
class Dog(Mammal):
    def bark(self):
        print("Dog bhaunk raha hai")

# Object creation
d = Dog()

# Parent classes ke features access karna
d.eat()   # Animal class ka feature
d.walk()  # Mammal class ka feature
d.bark()  # Dog class ka apna feature
```

**Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
# Parent Class 1
class Father:
    def hard_work(self):
        print("Father's Habit: Hard Working")

# Parent Class 2
class Mother:
    def cooking(self):
        print("Mother's Habit: Good Cooking")

# Child Class inheriting both parents
class Child(Father, Mother):
    def studying(self):
        print("Child's Habit: Studying")

# Object creation
c = Child()

# Accessing habits from both parents
c.hard_work()  # Father se inherit hua
c.cooking()    # Mother se inherit hua
c.studying()   # Child ka apna habit
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)  # Parent class ka constructor call karta hai
        self.roll_no = roll_no

# Usage
s = Student("Ahmed", 101)
print(f"Name: {s.name}, Roll No: {s.roll_no}")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
import math

# Parent Class
class Shape:
    def __init__(self, color="Red"):
        self.color = color

# Child Class 1
class Circle(Shape):
    def __init__(self, radius, color="Red"):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

# Child Class 2
class Square(Shape):
    def __init__(self, side, color="Red"):
        super().__init__(color)
        self.side = side

    def area(self):
        return self.side ** 2

# Testing
c = Circle(5)
s = Square(4)

print("Circle Area:", c.area(), "| Color:", c.color)
print("Square Area:", s.area(), "| Color:", s.color)
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Parent Class
class Bird:
    def fly(self):
        print("Most birds can fly")

# Child Class
class Ostrich(Bird):
    def fly(self):  # Parent class ke method ko override kar raha hai
        print("Ostriches cannot fly")

# Testing
b = Bird()
b.fly()  # Output: Most birds can fly

o = Ostrich()
o.fly()  # Output: Ostriches cannot fly
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
class Dog:
    def sound(self):
        return "Woof!"

class Cat:
    def sound(self):
        return "Meow!"

# Polymorphic function
def make_sound(animal_object):
    print(animal_object.sound())

# Usage
make_sound(Dog())  # Output: Woof!
make_sound(Cat())  # Output: Meow!

```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

# Usage
p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2  # Calls __add__ magic method
print("Result:", p3)  # Output: (6, 8)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

# Direct object creation raises TypeError
try:
    c = Computer()
except TypeError as e:
    print("Error:", e)
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):  # Concrete method
        print("Engine Started.")

    @abstractmethod
    def drive(self):  # Abstract method
        pass

class SportsCar(Car):
    def drive(self):
        print("Driving at high speed!")

# Usage
car = SportsCar()
car.start_engine()  # Concrete method call
car.drive()         # Abstract method call
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
from abc import ABC, abstractmethod

# Abstraction: Hiding implementation details
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")

# Encapsulation: Bundling data & protecting access
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private data

    def get_balance(self):
        return self.__balance

# Usage
Car().start()
acc = BankAccount(500)
print("Balance:", acc.get_balance())
```
